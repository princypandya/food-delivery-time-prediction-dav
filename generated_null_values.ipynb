```python
import pandas as pd
import numpy as np

# Load CSV
df = pd.read_csv("Food_Delivery_Time_Prediction.csv")

np.random.seed(42)
num_rows = len(df)

# Clear any existing nulls (optional)
df = df.fillna(df.mode().iloc[0])

# Nulls assignment as per conditions
# Order_ID: 8 null values
order_id_nulls = 8
order_id_null_indices = np.random.choice(num_rows, size=order_id_nulls, replace=False)
df.loc[order_id_null_indices, 'Order_ID'] = np.nan

# Customer_Location: 2 or 3 nulls
cust_loc_nulls = np.random.choice([2, 3])
cust_loc_null_indices = np.random.choice(list(set(range(num_rows)) - set(order_id_null_indices)), size=cust_loc_nulls, replace=False)
df.loc[cust_loc_null_indices, 'Customer_Location'] = np.nan

# Restaurant_Location: 2 or 3 nulls
rest_loc_nulls = np.random.choice([2, 3])
excluded_indices = set(order_id_null_indices).union(set(cust_loc_null_indices))
rest_loc_null_indices = np.random.choice(list(set(range(num_rows)) - excluded_indices), size=rest_loc_nulls, replace=False)
df.loc[rest_loc_null_indices, 'Restaurant_Location'] = np.nan

# For other columns: add minimum 5 and max 20 nulls randomly
other_columns = [col for col in df.columns if col not in ['Order_ID', 'Customer_Location', 'Restaurant_Location']]
used_indices = excluded_indices.union(set(rest_loc_null_indices))

for col in other_columns:
    available_indices = list(set(range(num_rows)) - used_indices)
    min_nulls = 5
    max_nulls = min(20, len(available_indices))
    n_nulls_col = np.random.randint(min_nulls, max_nulls + 1)
    null_indices = np.random.choice(available_indices, size=n_nulls_col, replace=False)
    df.loc[null_indices, col] = np.nan
    # Update used indices so no overlap
    used_indices = used_indices.union(set(null_indices))

# Save to CSV
df.to_csv("Food_Delivery_Time_Prediction_custom_nulls.csv", index=False)
```
